```python
import pandas as pd
import numpy as np
```

```python
num_of_nodes = 150
similarity_threshold = 0.9
```

### **Similarity Threshold**:

* Quantifying the similarity. 
* If the behavorial metrics goes above or equals 90%(`>=90`), they are more likely to be `Homophilic`.

```python
# Features: [Risk Tolerance, Observation Index, Impulsivity]

# Setting a baseline for different personality types
base_profiles = {
    0: [0.7, 0.2, 0.95], 
    1: [0.47, 0.9, 0.32], 
    2: [0.9, 0.87, 0.5]
}

# 0: Impulsive, 1: Analytical, 3: The Gamblers
```

```python
np.random.seed(42)
node_metrics = []
node_labels = []
```

```python
for _ in range(num_of_nodes):
    archetype = np.random.choice([0, 1, 2]) # For a specific arhcetype or baseline, choose profile 0, 1 or 2.
    noise = np.random.normal(0, 0.1, 3) # Injecting 'Gaussian' variance for true randomness
    
    metrics = np.clip(base_profiles[archetype] + noise, 0, 1) # (baseline profile +/- variance, lower limit, upper limit)
    
    node_metrics.append(metrics)
    node_labels.append(archetype)
```
