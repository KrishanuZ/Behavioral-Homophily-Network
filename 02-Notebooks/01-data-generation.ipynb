```python
import pandas as pd
import numpy as np
```

```python
num_of_nodes = 150
similarity_threshold = 0.9
```

### **Similarity Threshold**:

* Quantifying the similarity. 
* If the behavorial metrics goes above or equals 90%(`>=90`), they are more likely to be `Homophilic`.

```python
# Features: [Risk Tolerance, Observation Index, Impulsivity]

# Setting a baseline for different personality types
base_profiles = {
    0: [0.7, 0.2, 0.95], 
    1: [0.47, 0.9, 0.32], 
    2: [0.9, 0.71, 0.68]
}

# 0: Impulsive, 1: Analytical, 3: The Gamblers
```

```python
np.random.seed(42)
node_metrics = []
node_labels = []
```

```python
for _ in range(num_of_nodes):
    profile_type = np.random.choice([0, 1, 2]) # For a specific arhcetype or baseline, choose profile 0, 1 or 2.
    variance = np.random.normal(0, 0.1, 3) # Injecting 'Gaussian' variance(or noise) for true randomness
    variance = np.round(variance, 2)
    
    metrics = np.clip(base_profiles[profile_type] + variance, 0.00, 1.00) # (baseline profile +/- variance, lower limit, upper limit)
    
    node_metrics.append(metrics)
    node_labels.append(profile_type)
```

```python
node_metrics = np.array(node_metrics)
node_labels = np.array(node_labels)
```

```python
print(f"The generated data looks like:")
print(f"Label \tMetrics")

for i in range(0, len(node_labels)):
    print(f"{node_labels[i]}:      {node_metrics[i]}")
```

```python
# Saving it using the numpy library

np.save('../01-Dataset/node_metrics.npy', node_metrics)
np.save('../01-Dataset/node_labels.npy', node_labels)
```
