## We wire the `Nodes`(Link)

---

### Extract the created NumPy data:

```python
import numpy as np
```

```python
node_labels = np.load('../../01-Dataset/node_labels.npy')
node_metrics = np.load('../../01-Dataset/node_metrics.npy')
```

### Using cosine similarity:

```python
# %pip install networkx
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_mat = cosine_similarity(node_metrics) # Calculate Behavioral 'Homophily'
# print(type(similarity_mat))
```

### Add nodes first:

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.Graph()

num_nodes = len(node_labels)
# print(num_nodes)
similarity_threshold = 0.915 # A hardcoded threshold for now

# Adding nodes to the graph
for i in range(0, num_nodes):
    G.add_node(i, label=node_labels[i])
```

### Adding edges:

```python
# Wiring the nodes IFF threshold > 90%

for i in range(num_nodes):
    for j in range(i + 1, num_nodes): # No self loops
        
        if similarity_mat[i, j] >= similarity_threshold:
            G.add_edge(i, j)
```

### Plot the network:

```python
# Custom-building the legend

from matplotlib.lines import Line2D

profiles = [
    Line2D([0], [0], marker='o', color='w', label='Impulsive', markerfacecolor='green', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Analytical', markerfacecolor='purple', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Gambler', markerfacecolor='red', markersize=10),
]
```

```python
plt.figure(figsize=(12, 8))

pos = nx.spring_layout(G, seed=42)  # The engine which pulls the connected node together

color = [
    "#ff0000" if label == 0 else "#de00fc" if label == 1 else "#38F100" for label in node_labels
]

nx.draw_networkx_nodes(G, pos, node_color=color, node_size=100)
nx.draw_networkx_edges(G, pos, alpha=0.2)

plt.title('The Network', fontsize=16)

plt.legend(handles=profiles, loc='upper right', title='Profiles')
plt.tight_layout()
plt.savefig('Graph.png', dpi=300)
plt.show()
```
